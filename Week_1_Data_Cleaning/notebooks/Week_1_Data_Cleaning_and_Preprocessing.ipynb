## 3. Data Quality Assessment

The dataset was examined for common data-quality issues including missing values, placeholder values, duplicate records, inconsistent categorical labels, and unusual numerical values.

### 3.1 Data Types

The data types of all columns were inspected to distinguish numerical variables from categorical variables and to identify any potential type-related issues.

```python
print("===== DATA TYPES =====")
print(df.dtypes)

print("\n===== DUPLICATE ROWS =====")
print("Number of duplicate rows:", df.duplicated().sum())

print("\n===== UNIQUE VALUES =====")
for column in df.columns:
    print(f"{column}: {df[column].nunique()} unique values")
```

### 3.3 Categorical Value Inspection

The unique values and frequency distributions of categorical columns were inspected to identify unexpected categories, placeholder values, and inconsistencies.

Particular attention was given to `workclass`, `occupation`, `native-country`, and the `income` target.

```python
print("===== CATEGORICAL VALUE CHECK =====")

categorical_columns = df.select_dtypes(include="object").columns

for column in categorical_columns:
    print(f"\n--- {column} ---")
    print(df[column].value_counts(dropna=False).head(20))
```

### 3.4 Numerical Data Investigation

Numerical columns were examined using descriptive statistics, including minimum and maximum values, averages, and quartiles.

Potentially unusual values were investigated before deciding whether they represented actual errors. Extreme values were not automatically removed simply because they were statistically unusual.

```python
print("===== NUMERICAL SUMMARY =====")

numerical_columns = df.select_dtypes(include="number").columns

print(df[numerical_columns].describe().T)
```

### 3.5 Investigation of Unusual Numerical Values

Specific numerical ranges were investigated to determine whether extreme observations appeared to be invalid.

Examples included age, hours worked per week, and capital gains. Since unusual values were not automatically confirmed as errors, they were retained unless there was sufficient evidence that they were invalid.

```python
print("===== EXTREME / UNUSUAL NUMERICAL VALUES =====")

print("\nCapital-gain = 99999:")
print((df["capital-gain"] == 99999).sum())

print("\nCapital-loss > 0:")
print((df["capital-loss"] > 0).sum())

print("\nHours-per-week below 10:")
print((df["hours-per-week"] < 10).sum())

print("\nHours-per-week above 80:")
print((df["hours-per-week"] > 80).sum())

print("\nAge below 18:")
print((df["age"] < 18).sum())

print("\nAge above 80:")
print((df["age"] > 80).sum())
```

### 3.6 Missing-Value Pattern Analysis

The number of rows containing at least one missing or unknown value was calculated to understand the overall extent of incomplete records in the dataset.

```python
print("===== MISSING VALUE PATTERNS =====")

missing_mask = (
    df["workclass"].isna() |
    df["occupation"].isna() |
    df["native-country"].isna() |
    (df["workclass"] == "?") |
    (df["occupation"] == "?") |
    (df["native-country"] == "?")
)

print("Rows containing at least one missing/unknown value:", missing_mask.sum())

print("\nRows with no missing/unknown values:", (~missing_mask).sum())
```

### 3.7 Detailed Duplicate Inspection

The duplicate records were inspected before removal to confirm that the identified records represented exact duplicate observations.

```python
duplicates = df[df.duplicated(keep=False)]

print("Total duplicate rows:", len(duplicates))

duplicates.head(10)
```

## 4. Data Cleaning and Preprocessing

Based on the data-quality assessment, several preprocessing steps were performed. The cleaning process addressed confirmed data-quality issues while avoiding unnecessary removal of potentially valid observations.

The main preprocessing steps were:

1. Convert `?` placeholders to missing values.
2. Remove duplicate records.
3. Standardize the income labels.
4. Remove rows containing missing values.
5. Reset the dataset index.
6. Perform final data-quality verification.

```python
# ==========================================
# STEP 16 - DATA CLEANING
# ==========================================

# Create a copy so that the original raw dataframe remains unchanged
df_clean = df.copy()

print("Original dataset shape:", df_clean.shape)

# ------------------------------------------
# 1. Convert '?' placeholders to NaN
# ------------------------------------------

df_clean = df_clean.replace("?", pd.NA)

print("\nMissing values after converting '?' to NaN:")
print(df_clean.isna().sum())

# ------------------------------------------
# 2. Remove duplicate rows
# ------------------------------------------

duplicates_before = df_clean.duplicated().sum()

df_clean = df_clean.drop_duplicates()

print("\nDuplicate rows removed:", duplicates_before)
print("Shape after removing duplicates:", df_clean.shape)

# ------------------------------------------
# 3. Standardize income labels
# ------------------------------------------

df_clean["income"] = df_clean["income"].str.replace(".", "", regex=False)

print("\nIncome categories after standardization:")
print(df_clean["income"].value_counts())

# ------------------------------------------
# 4. Remove rows containing missing values
# ------------------------------------------

missing_rows_before = df_clean.isna().any(axis=1).sum()

df_clean = df_clean.dropna()

print("\nRows removed due to missing values:", missing_rows_before)
print("Final cleaned dataset shape:", df_clean.shape)

# ------------------------------------------
# 5. Reset index
# ------------------------------------------

df_clean = df_clean.reset_index(drop=True)

print("\nFinal index reset successfully.")

print("\nFinal cleaned dataset preview:")
display(df_clean.head())
```

```python
# Save the cleaned dataset

df_clean.to_csv("adult_cleaned.csv", index=False)

print("Cleaned dataset saved successfully.")
print("File name: adult_cleaned.csv")
print("Final shape:", df_clean.shape)
```

```python
print("===== CLEANING VERIFICATION =====")

print("\nFinal shape:")
print(df_clean.shape)

print("\nMissing values:")
print(df_clean.isna().sum())

print("\nDuplicate rows:")
print(df_clean.duplicated().sum())

print("\nIncome categories:")
print(df_clean["income"].value_counts())

print("\nUnique income labels:")
print(df_clean["income"].unique())
```

### 4.1 Final Duplicate Check

After standardizing the income labels, a second duplicate check was performed.

This was necessary because two records that previously had different textual income labels could become identical after removing the trailing period.

The remaining duplicate records were therefore removed before the final dataset was saved.

```python
# Remove duplicates created after income standardization

duplicates_remaining = df_clean.duplicated().sum()

print("Duplicates before final duplicate removal:", duplicates_remaining)

df_clean = df_clean.drop_duplicates().reset_index(drop=True)

print("Duplicates removed:", duplicates_remaining)
print("Final cleaned dataset shape:", df_clean.shape)
```

## 5. Final Data Quality Verification

The cleaned dataset was checked again after preprocessing.

The final verification examined:

- Dataset dimensions
- Missing values
- Duplicate records
- Income categories
- Unique income labels

The purpose of this step was to confirm that the cleaning operations produced the expected final dataset.

```python
print("===== FINAL DATA QUALITY CHECK =====")

print("\nFinal shape:")
print(df_clean.shape)

print("\nMissing values:")
print(df_clean.isna().sum().sum())

print("\nDuplicate rows:")
print(df_clean.duplicated().sum())

print("\nIncome categories:")
print(df_clean["income"].value_counts())

print("\nUnique income labels:")
print(df_clean["income"].unique())
```

## 6. Save the Cleaned Dataset

After the final quality checks confirmed that the dataset contained no missing values or duplicate records, the cleaned dataset was saved as `adult_cleaned.csv`.

This file will be used as the cleaned dataset for subsequent analysis and modeling tasks.

```python
df_clean.to_csv("adult_cleaned.csv", index=False)

print("Final cleaned dataset saved successfully.")
print("Shape:", df_clean.shape)
```
