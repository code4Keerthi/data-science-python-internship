## 2. Data Visualization

Visualizations are used to make important patterns and distributions in the dataset easier to understand.

The first visualization examines the distribution of the income categories in the cleaned dataset.

```python
plt.figure(figsize=(8, 5))
sns.countplot(data=df, x="income")
plt.title("Income Distribution")
plt.xlabel("Income Category")
plt.ylabel("Number of Records")
plt.tight_layout()
plt.savefig("/content/income_distribution.png", dpi=300)
plt.show()
```

### 2.1 Education Distribution

The education distribution shows the number of records belonging to each education category.

This visualization helps identify the most common education levels represented in the cleaned dataset.

```python
plt.figure(figsize=(10, 6))
sns.countplot(
    data=df,
    y="education",
    order=df["education"].value_counts().index
)
plt.title("Education Distribution")
plt.xlabel("Number of Records")
plt.ylabel("Education Level")
plt.tight_layout()
plt.savefig("/content/education_distribution.png", dpi=300)
plt.show()
```

### 2.2 Workclass Distribution

The workclass distribution shows how the records are distributed across different employment categories.

This visualization helps identify the most common types of employment represented in the dataset.

```python
plt.figure(figsize=(10, 6))
sns.countplot(
    data=df,
    y="workclass",
    order=df["workclass"].value_counts().index
)
plt.title("Workclass Distribution")
plt.xlabel("Number of Records")
plt.ylabel("Workclass")
plt.tight_layout()
plt.savefig("/content/workclass_distribution.png", dpi=300)
plt.show()
```

### 2.3 Age Distribution

The age distribution shows how the observations are distributed across different age values.

A histogram is used because age is a numerical variable. The visualization helps identify the general shape and spread of the age distribution.

```python
plt.figure(figsize=(10, 6))
sns.histplot(data=df, x="age", bins=20, kde=True)
plt.title("Age Distribution")
plt.xlabel("Age")
plt.ylabel("Number of Records")
plt.tight_layout()
plt.savefig("/content/age_distribution.png", dpi=300)
plt.show()
```

### 2.4 Hours Worked per Week Distribution

The hours-per-week distribution shows how working hours are distributed across the observations.

A histogram is used to examine the frequency and spread of weekly working hours in the cleaned dataset.

```python
plt.figure(figsize=(10, 6))
sns.histplot(data=df, x="hours-per-week", bins=20, kde=True)
plt.title("Hours Worked per Week Distribution")
plt.xlabel("Hours per Week")
plt.ylabel("Number of Records")
plt.tight_layout()
plt.savefig("/content/hours_per_week_distribution.png", dpi=300)
plt.show()
```

### 2.5 Income Distribution Across Education Levels

The relationship between education level and income category is examined using a grouped count plot.

The visualization compares the number of observations in each income category across different education levels. This helps identify differences in the distribution of income categories across education groups.

```python
plt.figure(figsize=(12, 7))
sns.countplot(
    data=df,
    y="education",
    hue="income",
    order=df["education"].value_counts().index
)
plt.title("Income Distribution Across Education Levels")
plt.xlabel("Number of Records")
plt.ylabel("Education Level")
plt.legend(title="Income")
plt.tight_layout()
plt.savefig("/content/income_across_education.png", dpi=300)
plt.show()
```

### 2.6 Hours Worked per Week by Income Category

The distribution of weekly working hours is compared across the two income categories using a box plot.

The box plot shows the median, spread, and potential extreme values of hours worked for each income group. This provides a comparison of working-hour distributions between the income categories.

```python
plt.figure(figsize=(8, 6))
sns.boxplot(
    data=df,
    x="income",
    y="hours-per-week"
)
plt.title("Hours Worked per Week by Income Category")
plt.xlabel("Income Category")
plt.ylabel("Hours per Week")
plt.tight_layout()
plt.savefig("/content/hours_by_income.png", dpi=300)
plt.show()
```

### 2.7 Age Distribution by Income Category

The age distribution is compared across the two income categories using a box plot.

This visualization helps compare the median age, spread, and potential extreme values between the income groups.

```python
plt.figure(figsize=(8, 6))
sns.boxplot(
    data=df,
    x="income",
    y="age"
)
plt.title("Age Distribution by Income Category")
plt.xlabel("Income Category")
plt.ylabel("Age")
plt.tight_layout()
plt.savefig("/content/age_by_income.png", dpi=300)
plt.show()
```

## 3. Correlation Analysis

Correlation analysis is used to examine the relationships between numerical variables.

A correlation coefficient ranges from -1 to +1. Values closer to +1 indicate a stronger positive linear relationship, while values closer to -1 indicate a stronger negative linear relationship. Values close to 0 indicate a weak linear relationship.

The correlation matrix is visualized using a heatmap to make relationships between the numerical variables easier to compare.

```python
numerical_columns = [
    "age",
    "fnlwgt",
    "education-num",
    "capital-gain",
    "capital-loss",
    "hours-per-week"
]

correlation_matrix = df[numerical_columns].corr()

print(correlation_matrix)
```

```python
plt.figure(figsize=(10, 7))
sns.heatmap(
    correlation_matrix,
    annot=True,
    fmt=".2f",
    cmap="coolwarm",
    center=0
)
plt.title("Correlation Matrix of Numerical Variables")
plt.tight_layout()
plt.savefig("/content/correlation_heatmap.png", dpi=300)
plt.show()
```

## 4. Grouped Analysis

Grouped analysis is used to compare the distribution of income categories across different categorical groups.

The first grouped analysis examines income categories within each workclass.

```python
income_by_workclass = pd.crosstab(
    df["workclass"],
    df["income"]
)

print(income_by_workclass)
```

### 4.1 Income Proportions by Workclass

Because workclass categories contain different numbers of observations, raw counts alone do not provide a complete comparison.

The income distribution is therefore converted into percentages within each workclass. This allows the income categories to be compared relative to the size of each workclass.

```python
income_by_workclass_percentage = pd.crosstab(
    df["workclass"],
    df["income"],
    normalize="index"
) * 100

print(income_by_workclass_percentage.round(2))
```

### 4.2 Income Proportions by Workclass Visualization

A stacked bar chart is used to compare the percentage distribution of the two income categories within each workclass.

Using percentages rather than raw counts makes the workclass categories more directly comparable despite differences in their total number of observations.

```python
income_by_workclass_percentage.plot(
    kind="bar",
    stacked=True,
    figsize=(10, 6)
)
plt.title("Income Proportions by Workclass")
plt.xlabel("Workclass")
plt.ylabel("Percentage of Records")
plt.legend(title="Income")
plt.xticks(rotation=45)
plt.tight_layout()
plt.savefig("/content/income_proportions_workclass.png", dpi=300)
plt.show()
```

### 4.3 Income Proportions by Education Level

The relationship between education level and income category is examined using proportions within each education group.

Calculating percentages within each education category allows the income distribution to be compared across education levels without being affected by differences in the number of observations in each group.

```python
income_by_education_percentage = pd.crosstab(
    df["education"],
    df["income"],
    normalize="index"
) * 100

print(income_by_education_percentage.round(2))
```

```python
income_by_education_percentage.plot(
    kind="bar",
    stacked=True,
    figsize=(12, 7)
)
plt.title("Income Proportions by Education Level")
plt.xlabel("Education Level")
plt.ylabel("Percentage of Records")
plt.legend(title="Income")
plt.xticks(rotation=45)
plt.tight_layout()
plt.savefig("/content/income_proportions_education.png", dpi=300)
plt.show()
```

## 5. EDA Findings and Observations

The exploratory analysis provided several observations about the structure and distribution of the cleaned Adult Income dataset.

### 5.1 Overall Dataset Characteristics

The cleaned dataset contains 45,175 records and 15 columns. The numerical variables show different ranges and distributions. The average age is approximately 38.56 years, while the average number of hours worked per week is approximately 40.94 hours.

The median number of hours worked per week is 40, indicating that 40 hours is a common value in the dataset.

### 5.2 Categorical Variable Distributions

The categorical analysis shows that `Private` is the largest workclass category, with 33,262 records. In the education variable, `HS-grad` is the most common category with 14,770 records, followed by `Some-college` with 9,887 records.

The `Married-civ-spouse` category is the largest marital-status group with 21,042 records. The `United-States` category dominates the native-country variable with 41,256 records.

The income variable contains 33,973 records in the `<=50K` category and 11,202 records in the `>50K` category.

### 5.3 Numerical Variable Distributions

The age histogram shows the distribution of ages across the dataset, while the hours-per-week histogram shows the distribution of weekly working hours.

The descriptive statistics show that age ranges from 17 to 90 years, while hours-per-week ranges from 1 to 99 hours.

The capital-gain variable has a median of 0 and a maximum of 99,999. Similarly, capital-loss has a median of 0 and a maximum of 4,356. This indicates that many observations have zero values for these two financial variables.

### 5.4 Correlation Analysis

The correlation analysis shows generally weak linear relationships among the numerical variables.

The strongest positive correlation in the numerical correlation matrix is between `education-num` and `hours-per-week`, with a correlation of approximately 0.146. The correlation between `education-num` and `capital-gain` is approximately 0.127, while the correlation between `age` and `hours-per-week` is approximately 0.102.

These correlation values are relatively small, so the numerical variables do not show strong linear relationships with one another in this analysis.

### 5.5 Income and Workclass

The grouped analysis shows differences in income proportions across workclass categories.

For example, the `Private` workclass contains 21.78% `>50K` observations, while `Self-emp-inc` contains 55.44% `>50K` observations.

These values describe the income distribution within each workclass in this dataset. They should not be interpreted as evidence that workclass directly causes differences in income.

### 5.6 Income and Education

The education-level analysis shows differences in the observed income proportions across education categories.

For example, the `Bachelors` category contains 42.02% `>50K` observations, while the `Masters` category contains 55.43% and the `Doctorate` category contains 73.35%.

The `Prof-school` category has 75.41% `>50K` observations. In comparison, the `HS-grad` category has 16.34% `>50K` observations.

These observations describe patterns in the dataset and do not establish a causal relationship between education and income.
